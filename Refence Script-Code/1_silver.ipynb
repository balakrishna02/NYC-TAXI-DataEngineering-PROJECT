##Reading data

###Trip type

```python
df_trip_type = spark.read.format('csv')\
                .option('inferSchema', True)\
                .option('header',True)\
                .load("abfss://bronze@nyctaxistoragechintu.dfs.core.windows.net/trip_type")
```

```python
df_trip_type.display()
```

###Trip Zone

```python
df_trip_zone =  spark.read.format('csv')\
                .option('inferSchema', True)\
                .option('header',True)\
                .load("abfss://bronze@nyctaxistoragechintu.dfs.core.windows.net/trip_zone")
```

```python
df_trip_zone.limit(10).display()
```

```python
from pyspark.sql.functions import *
from pyspark.sql.types import *
```

```python
df_trip = spark.read.format('parquet')\
                .schema(myschema)\
                .option('header',True)\
                .option('recursiveFileLookup',True)\
                .load("abfss://bronze@nyctaxistoragechintu.dfs.core.windows.net/trips2023data/")

```

```python
myschema='''
        VendorID BIGINT,
        tpep_pickup_datetime TIMESTAMP,
        tpep_dropoff_datetime TIMESTAMP,
        store_and_fwd_flag STRING,
        RatecodeID BIGINT,
        PULocationID BIGINT,
        DOLocationID BIGINT,
        passenger_count BIGINT,
        trip_distance DOUBLE,
        fare_amount DOUBLE,
        extra DOUBLE,
        mta_tax DOUBLE,
        tip_amount DOUBLE,
        tolls_amount DOUBLE,
        ehail_fee DOUBLE,
        improvement_surcharge DOUBLE,
        total_amount DOUBLE,
        payment_type BIGINT,
        trip_type BIGINT,
        congestion_surcharge DOUBLE
'''
```

```python
df_trip.limit(10).display()
```

##Data Transfromation

###Trip Type


```python
df_trip_type.display()
```

```python
df_trip_type = df_trip_type.withColumnRenamed('description',"trip_description")

df_trip_type.display()
```

```python
df_trip_type.write.format('parquet')\
                .mode('append')\
                .option('path','abfss://silver@nyctaxistoragechintu.dfs.core.windows.net/trip_type')\
                .save()
```

###Trip Zone


```python
df_trip_zone.display()
```

```python
df_trip_zone.withColumn('Zone1', split(col('Zone'), '/')).display()
                        #    .withColumn('Zone2', split(col('Zone'), '/')[1]).display()
```

```python

df_trip_zone = df_trip_zone.withColumn('Zone1', split(col('Zone'), '/')[0]) \
    .withColumn('Zone2', when(size(split(col('Zone'), '/')) > 1, split(col('Zone'), '/')[1]).otherwise(None))

df_trip_zone.display()
```

```python
df_trip_zone.write.format('parquet')\
    .mode('append')\
    .option('path','abfss://silver@nyctaxistoragechintu.dfs.core.windows.net/trip_zone')\
    .save()
```

###Trip

```python
df_trip.display()
```

```python
df_trip.withColumn('trip_year', year('tpep_pickup_datetime'))\
        .withColumn('trip_month', month('tpep_pickup_datetime'))\
        .withColumn('trip_day', to_date('tpep_pickup_datetime')).display()
```

```python
df_trip = df_trip.select('vendorID','PULocationID','DOLocationID','trip_distance','fare_amount','total_amount')

df_trip.display()
```

```python
df_trip.write.format('parquet')\
        .mode('append')\
        .option('path', 'abfss://silver@nyctaxistoragechintu.dfs.core.windows.net/trip')\
        .save()
        
```
